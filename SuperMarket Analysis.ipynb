```python
#task 1
#Mean is the average value of the data. Median is the middle value after sorting the data. Mode is the value that occurs most often. These measures help describe the centre of a dataset.
import pandas as pd
# Load the dataset
file_path = r"C:\Users\Dell\Downloads\SuperMarket Analysis.csv"
df = pd.read_csv(file_path)
# Select Total column
data = df["Quantity"]
# Calculate Mean, Median and Mode
mean = data.mean()
median = data.median()
mode = data.mode()[0]
# Display results
print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)



#task 2
#Variance measures how far values are spread from the mean. Standard deviation is the square root of variance and gives the spread in the same unit as the original data.
# Calculate variance and standard deviation
variance = data.var()
standard_deviation = data.std()
# Display results
print("Variance:", variance)
print("Standard Deviation:", standard_deviation)



#task 3
#Correlation measures the strength and direction of the relationship between two numerical variables. Probability measures the chance of an event happening.
# Calculate correlation
correlation = df["Quantity"].corr(df["Unit price"])
print("Correlation:", correlation)
# Determine relationship  
if correlation > 0.5:
    print("Relationship: Positive")
elif correlation < -0.5:
    print("Relationship: Negative")
else:
    print("Relationship: Weak")
# Find probability of Quantity greater than 5
favourable = (df["Quantity"] > 5).sum()
total = len(df)
probability = favourable / total
print("Favourable outcomes:", favourable)
print("Total outcomes:", total)
print("Probability:", probability)
print("Probability percentage:", probability * 100, "%")



#task4
#An outlier is a value that is unusually far from the rest of the data. The IQR method is commonly used to detect potential outliers.
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1
# Calculate limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
# Find outliers
outliers = data[(data < lower_limit) | (data > upper_limit)]
# Display results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:")
print(outliers)

```

```python
#task5
#Matplotlib is used to create basic statistical and business charts. The following code creates a line chart, bar chart, pie chart, histogram and scatter plot.
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
file_path = r"C:\Users\Dell\Downloads\SuperMarket Analysis.csv"
df = pd.read_csv(file_path)


# 1. LINE CHART
daily_sales = df.groupby("Date")["Sales"].sum()

plt.plot(daily_sales.index, daily_sales.values)
plt.title("Daily Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=90)
plt.show()


# 2. BAR CHART
product_sales = df.groupby("Product line")["Sales"].sum()

plt.bar(product_sales.index, product_sales.values)
plt.title("Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()


# 3. PIE CHART
payment = df["Payment"].value_counts()

plt.pie(payment.values,
        labels=payment.index,
        autopct="%1.1f%%")
plt.title("Payment Methods")
plt.show()


# 4. HISTOGRAM
plt.hist(df["Quantity"], bins=10)
plt.title("Distribution of Quantity")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.show()


# 5. SCATTER PLOT
plt.scatter(df["Quantity"], df["Sales"])
plt.title("Quantity vs Sales")
plt.xlabel("Quantity")
plt.ylabel("Sales")
plt.show()
```

```python
#task 6
#Seaborn provides attractive statistical visualizations and works well with pandas DataFrames
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
file_path = r"C:\Users\Dell\Downloads\SuperMarket Analysis.csv"
df = pd.read_csv(file_path)


# 1. COUNT PLOT
sns.countplot(data=df, x="Product line")
plt.title("Number of Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()


# 2. BOX PLOT
sns.boxplot(data=df, x="Product line", y="Sales")
plt.title("Sales Distribution by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()


# 3. HEATMAP
numeric_data = df[[
    "Unit price",
    "Quantity",
    "Tax 5%",
    "Sales",
    "cogs",
    "gross income",
    "Rating"
]]

correlation = numeric_data.corr()

sns.heatmap(correlation, annot=True)
plt.title("Correlation Heatmap")
plt.show()


# 4. PAIR PLOT
pair_data = df[[
    "Unit price",
    "Quantity",
    "Sales",
    "gross income",
    "Rating"
]]

sns.pairplot(pair_data)
plt.show()
```

```python
#Dataset After Cleaning
#Data Inspection means understanding the structure and contents of the dataset by checking its rows, columns, column names, data types, statistical summary, and missing values. 
#Data Cleaning means fixing problems such as missing values, duplicate records, and incorrect or inconsistent data.
import pandas as pd

# Load dataset
file_path = r"C:\Users\Dell\Downloads\SuperMarket Analysis.csv"
df = pd.read_csv(file_path)

# ---------------- DATA INSPECTION ----------------

# Number of rows and columns
print("Rows and Columns:")
print(df.shape)

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Data types
print("\nData Types:")
print(df.dtypes)

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())


# ---------------- DATA CLEANING ----------------

# Remove duplicate records
df = df.drop_duplicates()

# Fill missing numerical values with median
numeric_columns = df.select_dtypes(include="number").columns
df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].median()
)

# Fill missing categorical values with mode
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

# Check missing values after cleaning
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

# Check duplicates after cleaning
print("\nDuplicate Records After Cleaning:")
print(df.duplicated().sum())

# Display cleaned dataset
print("\nCleaned Dataset:")
print(df.head())
```

```python
#task 8
#Correlation analysis helps identify the relationship between numerical variables. A correlation value ranges from -1 to +1.
# +1 → Strong positive relationship
# -1 → Strong negative relationship
# 0 → Little or no linear relationship

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
file_path = r"C:\Users\Dell\Downloads\SuperMarket Analysis.csv"
df = pd.read_csv(file_path)

# Select numerical columns
numeric_data = df[[
    "Unit price",
    "Quantity",
    "Tax 5%",
    "Sales",
    "cogs",
    "gross income",
    "Rating"
]]

# Calculate correlation
correlation = numeric_data.corr()

print("Correlation Matrix:")
print(correlation)

# Display heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Correlation Analysis of Supermarket Data")
plt.show()
```
